```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

from matplotlib.lines import Line2D

plt.rcParams.update({'font.size': 10})
```

```python
def calculate_temp_variation(coefficients, model_name):

    temp_coef = []
    for i, coefs in coefficients.iterrows():

        temp = np.linspace(10, 35, 200)

        if model_name == 'Year + Month + Climate(lag 1)':
            # More readable version
            degree1_terms = sum(temp * coefs[f'mean_2m_air_temp_degree1_lag{i}'] for i in range(1, 2))
            degree2_terms = sum((temp ** 2) * coefs[f'mean_2m_air_temp_degree2_lag{i}'] for i in range(1, 2))

            temp_variation = pd.DataFrame({
                'coef': degree1_terms + degree2_terms, #+ degree3_terms,
                'temp': temp
            })
        elif model_name in ['Year + Month + Climate(lag 1-5)', 'Mexico', 'Peru', 'Brazil + Mexico + Peru']:

            # More readable version
            degree1_terms = sum(temp * coefs[f'mean_2m_air_temp_degree1_lag{i}'] for i in range(1, 6))
            degree2_terms = sum((temp ** 2) * coefs[f'mean_2m_air_temp_degree2_lag{i}'] for i in range(1, 6))

            temp_variation = pd.DataFrame({
                'coef': degree1_terms + degree2_terms, #+ degree3_terms,
                'temp': temp
            })
        elif model_name in ['Childs et al. (2025)']:
            degree1_terms = sum(temp * coefs[f'mean_2m_air_temp_degree1_lag{i}'] for i in range(1, 4))
            degree2_terms = sum((temp ** 2) * coefs[f'mean_2m_air_temp_degree2_lag{i}'] for i in range(1, 4))
            degree3_terms = sum((temp ** 3) * coefs[f'mean_2m_air_temp_degree3_lag{i}'] for i in range(1, 4))

            temp_variation = pd.DataFrame({
                'coef': degree1_terms + degree2_terms + degree3_terms,
                'temp': temp
            })
        else:
            # More readable version
            degree1_terms = sum(temp * coefs[f'mean_2m_air_temp_degree1_lag{i}'] for i in range(1, 4))
            degree2_terms = sum((temp ** 2) * coefs[f'mean_2m_air_temp_degree2_lag{i}'] for i in range(1, 4))

            temp_variation = pd.DataFrame({
                'coef': degree1_terms + degree2_terms, #+ degree3_terms,
                'temp': temp
            })

        temp_variation['model'] = i
        temp_coef.append(temp_variation)


    temp_coef = pd.concat(temp_coef)
    temp_coef_variation = temp_coef.groupby('model').apply(lambda x: x.set_index('temp')['coef'].diff()).reset_index().transpose()
    temp_coef_variation = temp_coef_variation.drop('model')

    quantiles = temp_coef_variation.quantile([0.025, 0.5, 0.975], axis=1).transpose().reset_index()
    quantiles['temp'] = quantiles['temp'].astype(float)

    return quantiles

models = [
    {'filename': 'data/standard_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate', 'color': 'k'},    
    {'filename': 'data/childs_coef_state_blockboot1000.csv', 'name': 'Childs et al. (2025)', 'color': 'k'},
    {'filename': 'data/lag1_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate(lag 1)', 'color': 'k'},
    {'filename': 'data/lag_45_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate(lag 1-5)', 'color': 'k'},
    {'filename': 'data/with_immunity_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate + Immunity', 'color': 'k'},
    {'filename': 'data/socioeconomic_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate + Socioeconomic', 'color': 'k'},
    {'filename': 'data/main_coef_state_mexico_blockboot1000.csv', 'name': 'Mexico', 'color': 'C4'},
    {'filename': 'data/main_coef_state_peru_blockboot1000.csv', 'name': 'Peru', 'color': 'C4'},
    {'filename': 'data/main_coef_state_all_blockboot1000.csv', 'name': 'Brazil + Mexico + Peru', 'color': 'C4'}
]

df = pd.read_csv('data/model_input_brazil_immunity_city.csv', parse_dates=['date_first_symptoms'])

fig, axarr = plt.subplots(3, 3, figsize=(10, 8.5), sharex=True, sharey=True)

for model, ax in zip(models, axarr.flatten()):

    coefficients_model = pd.read_csv(model['filename'])
    quantiles_model = calculate_temp_variation(coefficients_model, model['name'])

    ax.axvspan(18, 34, color='C1', alpha=0.2)
    ax.axvspan(26, 29, color='C3', alpha=0.2)

    print(model, quantiles_model.set_index('temp')[0.5].abs().idxmin())

    quantiles_model.plot(x='temp', y=0.5, ax=ax, color=model['color'], legend=False)
    ax.fill_between(
        quantiles_model['temp'],
        quantiles_model[0.025],
        quantiles_model[0.975],
        color=model['color'], alpha=0.2
    )

    ax.axhline(0, color='k', ls='--')
    ax.set_xlim(df['mean_2m_air_temp_degree1'].quantile(0.01), df['mean_2m_air_temp_degree1'].quantile(0.99))
    ax.set_xlabel('Temperature (°C)')
    ax.set_ylim([-0.3, 0.3])
    ax.set_title(model['name'], fontsize=10)

    legend_elements = []
    legend_elements = [
        Line2D([0], [0], color=model['color'], lw=2, label='Temperature Coeff. Variation'),
    ]

    # Place legend on the figure level or first axis
    ax.legend(handles=legend_elements, frameon=False, fontsize=8, loc='lower left')

axarr[0, 0].set_ylabel('Temp. Coefficient Variation')
axarr[1, 0].set_ylabel('Temp. Coefficient Variation')
axarr[2, 0].set_ylabel('Temp. Coefficient Variation')

plt.tight_layout()
plt.savefig('img/temp_coef_variation_models.pdf', dpi=300, bbox_inches='tight')
```

```python
act = pd.read_csv('data/sens_incidence_results_all_3150_all_ensembles.csv')
nat = pd.read_csv('data/sens_incidence_results_all_3150_nat_ensembles.csv')

regions = ['North', 'Northeast', 'Middle-West', 'Southeast', 'South']

act = act[act['region'].isin(regions)]
nat = nat[nat['region'].isin(regions)]

act = act.groupby([
    'model', 'region', 'bootstrap_iteration', 'ensemble'
]).apply(lambda x: pd.Series({
    'predicted_incidence': x['predicted_incidence'].sum(),
    'population': x['population'].sum(),
    'fitted_incidence': x['fitted_incidence'].sum(),
    'fitted_population': x['fitted_population'].sum(),
}), include_groups=False).reset_index()

nat = nat.groupby([
    'model', 'region', 'bootstrap_iteration', 'ensemble'
]).apply(lambda x: pd.Series({
    'predicted_incidence': x['predicted_incidence'].sum(),
    'population': x['population'].sum(),
    'fitted_incidence': x['fitted_incidence'].sum(),
    'fitted_population': x['fitted_population'].sum(),
}), include_groups=False).reset_index()

act['normed_incidence'] = act['predicted_incidence'] / act['fitted_incidence']
nat['normed_incidence'] = nat['predicted_incidence'] / nat['fitted_incidence']

act = act.groupby(['model', 'region', 'bootstrap_iteration']).mean().reset_index()
nat = nat.groupby(['model', 'region', 'bootstrap_iteration']).mean().reset_index()
```

```python
plt.rcParams.update({'font.size': 14})

diff = act.set_index(
    ['model', 'region', 'bootstrap_iteration']
)[['predicted_incidence', 'fitted_incidence']].rename(
    columns={'predicted_incidence': 'act', 'fitted_incidence': 'fitted'}
).join(
    nat.set_index(
        ['model', 'region', 'bootstrap_iteration']
    )[['predicted_incidence']].rename(
        columns={'predicted_incidence': 'nat'}  
    )
)

diff['diff'] = 100 * (diff['act'] - diff['nat']) / diff['fitted']
diff = diff.reset_index()

model_names = { 
    'childs': 'Childs et al. (2025)', 
    'lag1': 'Year + Month + Climate(lag 1)', 
    'lag45': 'Year + Month + Climate(lag 1-5)', 
    'standard': 'Year + Month + Climate', 
    'immunity': 'Year + Month + Climate + Immunity', 
    'socioeconomic': 'Year + Month + Climate + Socioeconomic',
}
diff['model'] = diff['model'].map(model_names)
diff = diff.sort_values('model')

fig, ax = plt.subplots(figsize=(10,6))

sns.boxplot(
    x='diff',
    y='region',
    hue='model',
    data=diff,
    ax=ax,
    gap=.2,
    order=['North', 'Northeast', 'Middle-West', 'Southeast', 'South'],
    palette={
        'Year + Month + Climate(lag 1-5)': 'C1',
        'Childs et al. (2025)': '#E0E0E0', 
        'Year + Month + Climate': '#C0C0C0', 
        'Year + Month + Climate + Immunity': '#808080', 
        'Year + Month + Climate(lag 1)': '#202020',
        'Year + Month + Climate + Socioeconomic': '#404040',
    }
)

# Move the legend outside the plot (to the right)
ax.legend(
    bbox_to_anchor=(1.05, 1),  # Shift to the right of the axis
    loc='upper left',
    borderaxespad=0.,
    frameon=False,
    fontsize=10
)

# Optional improvements
ax.set_xlabel("Normalized Attribution Signal (ACT - NAT) (%)")
ax.set_ylabel("Region")
ax.axvline(0, color='k', ls='--')  # Vertical line at 0 for reference
plt.tight_layout()  # Adjust layout to make room for legend
plt.savefig('img/attribution_signal_models_region_boxplot.pdf', dpi=300, bbox_inches='tight')
```
