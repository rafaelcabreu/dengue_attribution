```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

from matplotlib.lines import Line2D

plt.rcParams.update({'font.size': 10})
```

```python
def calculate_temp_variation(coefficients, model_name):

    bsplines = pd.read_csv('data/temperature_bsplines.csv')

    temp_coef = []
    for i, coefs in coefficients.iterrows():

        temp = np.linspace(10, 35, 200)

        if model_name == 'Year + Month + Climate(lag 1)':
            lag1_terms = sum(bsplines[f'bs_{i}'] * coefs.loc[f'temp_bs_lag1{i}'] for i in range(1, 5))

            temp_variation = pd.DataFrame({
                'coef': lag1_terms,
                'temp': temp
            })
        elif model_name in ['Year + Month + Climate(lag 1-5)', 'Mexico', 'Peru', 'Brazil + Mexico + Peru']:
            lag1_terms = sum(bsplines[f'bs_{i}'] * coefs.loc[f'temp_bs_lag1{i}'] for i in range(1, 5))
            lag2_terms = sum(bsplines[f'bs_{i}'] * coefs.loc[f'temp_bs_lag2{i}'] for i in range(1, 5))
            lag3_terms = sum(bsplines[f'bs_{i}'] * coefs.loc[f'temp_bs_lag3{i}'] for i in range(1, 5))
            lag4_terms = sum(bsplines[f'bs_{i}'] * coefs.loc[f'temp_bs_lag4{i}'] for i in range(1, 5))
            lag5_terms = sum(bsplines[f'bs_{i}'] * coefs.loc[f'temp_bs_lag5{i}'] for i in range(1, 5))

            temp_variation = pd.DataFrame({
                'coef': lag1_terms + lag2_terms + lag3_terms + lag4_terms + lag5_terms,
                'temp': temp
            })
        elif model_name in ['Year + Month + Climate(No regional terms)']:
            degree1_terms = sum(temp * coefs.loc[f'mean_2m_air_temp_degree1_lag{i}'] for i in range(1, 4))
            degree2_terms = sum((temp ** 2) * coefs.loc[f'mean_2m_air_temp_degree2_lag{i}'] for i in range(1, 4))
            degree3_terms = sum((temp ** 3) * coefs.loc[f'mean_2m_air_temp_degree3_lag{i}'] for i in range(1, 4))

            temp_variation = pd.DataFrame({
                'coef': degree1_terms + degree2_terms + degree3_terms,
                'temp': temp
            })
        else:
            lag1_terms = sum(bsplines[f'bs_{i}'] * coefs.loc[f'temp_bs_lag1{i}'] for i in range(1, 5))
            lag2_terms = sum(bsplines[f'bs_{i}'] * coefs.loc[f'temp_bs_lag2{i}'] for i in range(1, 5))
            lag3_terms = sum(bsplines[f'bs_{i}'] * coefs.loc[f'temp_bs_lag3{i}'] for i in range(1, 5))

            temp_variation = pd.DataFrame({
                'coef': lag1_terms + lag2_terms + lag3_terms,
                'temp': temp
            })

        temp_variation['model'] = i
        temp_coef.append(temp_variation)

    temp_coef = pd.concat(temp_coef)
    temp_coef_variation = temp_coef.groupby('model').apply(lambda x: x.set_index('temp')['coef'].diff()).reset_index().transpose()
    temp_coef_variation = temp_coef_variation.drop('model')

    quantiles = temp_coef_variation.quantile([0.025, 0.5, 0.975], axis=1).transpose().reset_index()
    quantiles['temp'] = quantiles['temp'].astype(float)

    return quantiles

models = [
    {'filename': 'data/standard_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month + Climate'},    
    {'filename': 'data/childs_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate(No regional terms)', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate(No regional terms)'},
    {'filename': 'data/lag1_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate(lag 1)', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate(lag 1)'},
    {'filename': 'data/lag_45_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate(lag 1-5)', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate(lag 1-5)'},
    {'filename': 'data/with_immunity_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate + Immunity', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate + Immunity'},
    {'filename': 'data/socioeconomic_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate + Socioeconomic', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate + Socioeconomic'},
    {'filename': 'data/lag1_cases_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate + Cases(lag1)', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate + Cases(lag1)'},    
    {'filename': 'data/glmm_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate(lag1-5);GLMM', 'color': 'C3', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate(lag1-5);GLMM'},    
    {'filename': 'data/main_coef_state_mexico_blockboot1000.csv', 'name': 'Mexico', 'color': 'C4', 'histfile': 'data/model_input_mexico_immunity_city.csv', 'title': 'Mexico'},
    {'filename': 'data/main_coef_state_all_blockboot1000.csv', 'name': 'Brazil + Mexico', 'color': 'C4', 'histfile': 'data/model_input_mexico-peru-brazil_immunity_city.csv', 'title': 'Brazil + Mexico + Peru'},
]

df = pd.read_csv('data/model_input_brazil_immunity_city.csv', parse_dates=['date_first_symptoms'])

fig, axarr = plt.subplots(2, 5, figsize=(12, 6), sharex=True, sharey=True)

for i, (model, ax) in enumerate(zip(models, axarr.flatten())):

    coefficients_model = pd.read_csv(model['filename'])
    quantiles_model = calculate_temp_variation(coefficients_model, model['name'])

    ax.axvspan(18, 34, color='C1', alpha=0.2)
    ax.axvspan(26, 29, color='C3', alpha=0.2)

    print(model, quantiles_model.set_index('temp')[0.5].abs().idxmin())

    quantiles_model.plot(x='temp', y=0.5, ax=ax, color=model['color'], legend=False)
    ax.fill_between(
        quantiles_model['temp'],
        quantiles_model[0.025],
        quantiles_model[0.975],
        color=model['color'], alpha=0.2
    )

    axt = ax.twinx()
    df_hist = pd.read_csv(model['histfile'])
    sns.histplot(df_hist['mean_2m_air_temp_degree1'], bins=np.arange(10,31,1), stat='density', color='gray', ax=axt)

    ax.axhline(0, color='k', ls='--')
    ax.set_xlim(df['mean_2m_air_temp_degree1'].quantile(0.01), df['mean_2m_air_temp_degree1'].quantile(0.99))
    ax.set_ylim([-1.2, 0.5])
    ax.set_title(model['title'], fontsize=10)
    ax.axhline(-0.6, color='k', ls='-')
    ax.set_yticks([-0.5, -0.25, 0.0, 0.25, 0.5])
    axt.set_ylim([0, 0.5])
    axt.set_yticks([0, 0.05, 0.1])
    axt.set_ylabel('')

    if i in [0, 5]:
        ax.text(7.5, -0.8, 'Temperature Coefficient \n Variation', color='k', rotation=90, ha='center', fontsize=12)

    if i in [4, 9]:
        axt.text(34, 0, 'Frequency', color='k', rotation=90, fontsize=12)
    else:
        axt.set_yticklabels([])

    if i == 7:
        ax.set_xlabel('Temperature (°C)')
    else:
        ax.set_xlabel('')
    

plt.tight_layout()
plt.savefig('img/temp_coef_variation_models.pdf', dpi=300, bbox_inches='tight')
```

```python
def calculate_prec_variation(coefficients, model_name):

    prec = np.linspace(0, 1200, 400)

    prec_coef = []
    for i, coefs in coefficients.iterrows():

        if model_name == 'Year + Month + Climate(lag 1)':
            lag1_terms = prec * coefs.loc[f'total_precipitation_lag1']

            prec_variation = pd.DataFrame({
                'coef': lag1_terms,
                'prec': prec
            })
        elif model_name in ['Year + Month + Climate(lag 1-5)', 'Mexico', 'Peru', 'Brazil + Mexico']:
            lag1_terms = prec * coefs.loc[f'total_precipitation_lag1']
            lag2_terms = prec * coefs.loc[f'total_precipitation_lag2']
            lag3_terms = prec * coefs.loc[f'total_precipitation_lag3']
            lag4_terms = prec * coefs.loc[f'total_precipitation_lag4']
            lag5_terms = prec * coefs.loc[f'total_precipitation_lag5']

            prec_variation = pd.DataFrame({
                'coef': lag1_terms + lag2_terms + lag3_terms + lag4_terms + lag5_terms,
                'prec': prec
            })
        else:
            lag1_terms = prec * coefs.loc[f'total_precipitation_lag1']
            lag2_terms = prec * coefs.loc[f'total_precipitation_lag2']
            lag3_terms = prec * coefs.loc[f'total_precipitation_lag3']

            prec_variation = pd.DataFrame({
                'coef': lag1_terms + lag2_terms + lag3_terms,
                'prec': prec
            })

        prec_variation['model'] = i
        prec_coef.append(prec_variation)

    prec_coef = pd.concat(prec_coef)
    prec_coef_variation = prec_coef.groupby('model').apply(lambda x: x.set_index('prec')['coef'].diff()).reset_index().transpose()
    prec_coef_variation = prec_coef_variation.drop('model')

    quantiles = prec_coef_variation.quantile([0.025, 0.5, 0.975], axis=1).transpose().reset_index()
    quantiles['prec'] = quantiles['prec'].astype(float)

    return quantiles

models = [
    {'filename': 'data/standard_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month + Climate'},    
    {'filename': 'data/childs_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate(No regional terms)', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate(No regional terms)'},
    {'filename': 'data/lag1_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate(lag 1)', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate(lag 1)'},
    {'filename': 'data/lag_45_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate(lag 1-5)', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate(lag 1-5)'},
    {'filename': 'data/with_immunity_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate + Immunity', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate + Immunity'},
    {'filename': 'data/socioeconomic_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate + Socioeconomic', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate + Socioeconomic'},
    {'filename': 'data/lag1_cases_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate + Cases(lag1)', 'color': 'k', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate + Cases(lag1)'},    
    {'filename': 'data/glmm_coef_state_blockboot1000.csv', 'name': 'Year + Month + Climate(lag1-5);GLMM', 'color': 'C3', 'histfile': 'data/model_input_brazil_immunity_city.csv', 'title': 'Year + Month +\nClimate(lag1-5);GLMM'},    
    {'filename': 'data/main_coef_state_mexico_blockboot1000.csv', 'name': 'Mexico', 'color': 'C4', 'histfile': 'data/model_input_mexico_immunity_city.csv', 'title': 'Mexico'},
    {'filename': 'data/main_coef_state_all_blockboot1000.csv', 'name': 'Brazil + Mexico', 'color': 'C4', 'histfile': 'data/model_input_mexico-peru-brazil_immunity_city.csv', 'title': 'Brazil + Mexico + Peru'},
]

df = pd.read_csv('data/model_input_brazil_immunity_city.csv', parse_dates=['date_first_symptoms'])

fig, axarr = plt.subplots(2, 5, figsize=(12, 6), sharex=True, sharey=True)

for i, (model, ax) in enumerate(zip(models, axarr.flatten())):

    coefficients_model = pd.read_csv(model['filename'])
    quantiles_model = calculate_prec_variation(coefficients_model, model['name'])

    quantiles_model.plot(x='prec', y=0.5, ax=ax, color=model['color'], legend=False)
    ax.fill_between(
        quantiles_model['prec'],
        quantiles_model[0.025],
        quantiles_model[0.975],
        color=model['color'], alpha=0.2
    )

    axt = ax.twinx()
    df_hist = pd.read_csv(model['histfile'])
    sns.histplot(df_hist['total_precipitation'], stat='density', color='gray', ax=axt, bins=np.linspace(0, 1200, 200))

    ax.axhline(0, color='k', ls='--')
    ax.set_xlim(df['total_precipitation'].min(), df['total_precipitation'].quantile(0.99))
    ax.set_ylim([-0.16, 0.02])
    ax.set_title(model['title'], fontsize=10)
    ax.axhline(-0.03, color='k', ls='-')
    ax.set_yticks([-0.02, 0.0, 0.02, 0.04])
    axt.set_ylim([0, 0.04])
    axt.set_yticks([0, 0.005, 0.01])
    axt.set_ylabel('')

    if i in [0, 5]:
        ax.text(-150, -0.10, 'Precipitation Coefficient \n Variation', color='k', rotation=90, ha='center', fontsize=12)

    if i in [4, 9]:
        axt.text(550, 0, 'Frequency', color='k', rotation=90, fontsize=12)
    else:
        axt.set_yticklabels([])

    if i == 7:
        ax.set_xlabel('Precipitation (mm)')
    else:
        ax.set_xlabel('')
    


plt.tight_layout()
plt.savefig('img/prec_coef_variation_models.pdf', dpi=300, bbox_inches='tight')
```

```python
act = pd.read_csv('data/sens_incidence_results_all_525_all_ensembles_v3.csv')
nat = pd.read_csv('data/sens_incidence_results_nat_525_all_ensembles_v3.csv')

regions = ['North', 'Northeast', 'Middle-West', 'Southeast', 'South']

act = act[act['region'].isin(regions)]
nat = nat[nat['region'].isin(regions)]

act = act.groupby([
    'model', 'region', 'bootstrap_iteration', 'ensemble'
]).apply(lambda x: pd.Series({
    'predicted_incidence': x['predicted_incidence'].sum(),
    'population': x['population'].sum(),
    'fitted_incidence': x['fitted_incidence'].sum(),
    'fitted_population': x['fitted_population'].sum(),
}), include_groups=False).reset_index()

nat = nat.groupby([
    'model', 'region', 'bootstrap_iteration', 'ensemble'
]).apply(lambda x: pd.Series({
    'predicted_incidence': x['predicted_incidence'].sum(),
    'population': x['population'].sum(),
    'fitted_incidence': x['fitted_incidence'].sum(),
    'fitted_population': x['fitted_population'].sum(),
}), include_groups=False).reset_index()

act['normed_incidence'] = act['predicted_incidence'] / act['fitted_incidence']
nat['normed_incidence'] = nat['predicted_incidence'] / nat['fitted_incidence']

act = act.groupby(['model', 'region', 'bootstrap_iteration']).mean().reset_index()
nat = nat.groupby(['model', 'region', 'bootstrap_iteration']).mean().reset_index()
```

```python
plt.rcParams.update({'font.size': 14})

diff = act.set_index(
    ['model', 'region', 'bootstrap_iteration']
)[['predicted_incidence', 'fitted_incidence']].rename(
    columns={'predicted_incidence': 'act', 'fitted_incidence': 'fitted'}
).join(
    nat.set_index(
        ['model', 'region', 'bootstrap_iteration']
    )[['predicted_incidence']].rename(
        columns={'predicted_incidence': 'nat'}  
    )
)

diff['diff'] = 100 * (diff['act'] - diff['nat']) / diff['fitted']
diff = diff.reset_index()

model_names = { 
    'childs': 'Year + Month + Climate(No regional terms)', 
    'lag1': 'Year + Month + Climate(lag 1)', 
    'lag45': 'Year + Month + Climate(lag 1-5)\n(selected)', 
    'standard': 'Year + Month + Climate', 
    'immunity': 'Year + Month + Climate + Immunity', 
    'socioeconomic': 'Year + Month + Climate + Socioeconomic',
}
diff['model'] = diff['model'].map(model_names)
diff = diff.sort_values('model')
diff['region'] = diff['region'].replace({'Middle-West': 'Centre-West'})
print(diff)

fig, ax = plt.subplots(figsize=(12,5))

sns.boxplot(
    y='diff',
    x='region',
    hue='model',
    data=diff,
    ax=ax,
    gap=.2,
    order=['North', 'Northeast', 'Centre-West', 'Southeast', 'South'],
    palette={
        'Year + Month + Climate(lag 1-5)\n(selected)': 'C1',
        'Year + Month + Climate(No regional terms)': '#E0E0E0', 
        'Year + Month + Climate': '#C0C0C0', 
        'Year + Month + Climate + Immunity': '#808080', 
        'Year + Month + Climate(lag 1)': '#202020',
        'Year + Month + Climate + Socioeconomic': '#404040',
    },
   showfliers=False,
   whis=(2.5, 97.5)
)

# Move the legend outside the plot (to the right)
ax.legend(
    #bbox_to_anchor=(1.05, 1),  # Shift to the right of the axis
    loc='lower right',
    borderaxespad=0.,
    frameon=False,
    fontsize=10
)

# Optional improvements
ax.set_ylabel("Normalized Attribution Signal\n(ACT - NAT) (%)")
ax.set_xlabel("Region")
ax.axhline(0, color='k', ls='--')  # Vertical line at 0 for reference
plt.tight_layout()  # Adjust layout to make room for legend
plt.savefig('img/attribution_signal_models_region_boxplot.pdf', dpi=300, bbox_inches='tight')
```
